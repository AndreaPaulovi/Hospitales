```python
import pandas as pd

# Importar el archivo CSV
df_urgencias = pd.read_csv("urgencias_hospitales.csv")

# Mostrar las primeras filas
print(df_urgencias.head())

# Verificar estructura
print(df_urgencias.info())
```

```python
# Seleccionar una sola columna como Series
nivel_triage = df_urgencias["Nivel_Triage"]

print("Nivel de Triage:")
print(nivel_triage)

# Seleccionar varias columnas
vista_hospitalaria = df_urgencias[
    ["Hospital", "Especialidad", "Tiempo_Espera_Min"]
]

print("\nVista hospitalaria:")
print(vista_hospitalaria)
```

```python
registro_75 = df_urgencias.loc[75]

print("Registro con índice 75:")
print(registro_75)
```

```python
duplicados = df_urgencias.duplicated().sum()

print("Cantidad de registros duplicados:", duplicados)
```

```python
df_urgencias = df_urgencias.drop_duplicates()

print("Duplicados después de eliminar:", df_urgencias.duplicated().sum())
```

```python
adultos_mayores = df_urgencias[
    df_urgencias["Edad_Paciente"] > 65
]

print("Pacientes mayores de 65 años:")
print(adultos_mayores)
```

```python
hospitales_seleccionados = df_urgencias[
    df_urgencias["Hospital"].isin([
        "Hospital Santa Sofía",
        "Clínica San Marcel"
    ])
]

print("Pacientes de los hospitales seleccionados:")
print(hospitales_seleccionados)
```

```python
espera_30_90 = df_urgencias[
    df_urgencias["Tiempo_Espera_Min"].between(30.0, 90.0)
]

print("Pacientes con espera entre 30 y 90 minutos:")
print(espera_30_90)
```

```python
triage_caro = df_urgencias.query(
    "Nivel_Triage == 'Triage I' and Costo_Procedimiento > 300000"
)

print("Triage I con costo superior a $300.000:")
print(triage_caro)
```

```python
filtro_compuesto = (
    (df_urgencias["Destino_Alta"] == "Hospitalización")
    & (df_urgencias["Especialidad"].str.startswith("M"))
)

resultado = df_urgencias[filtro_compuesto]

print("Hospitalización y especialidad que comienza por M:")
print(resultado)
```

```python
promedio_espera = df_urgencias["Tiempo_Espera_Min"].mean()

print("Promedio de espera:", promedio_espera)
```

```python
df_urgencias["Tiempo_Espera_Min"] = (
    df_urgencias["Tiempo_Espera_Min"].fillna(promedio_espera)
)
```
